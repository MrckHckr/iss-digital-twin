```python
import pandas as pd
import numpy as np

n = 10000

df = pd.DataFrame({
    "temperature": np.random.normal(22,2,n),
    "power_usage": np.random.normal(120,15,n),
    "battery_level": np.random.normal(90,5,n),
    "vibration": np.random.normal(0.02,0.005,n)
})

df.head()
```

```python
modules = [
    "Node1",
    "Node2",
    "Harmony",
    "Columbus",
    "Kibo",
    "Cupola"
]
```

```python
iss_modules = pd.DataFrame({
    "module": modules,
    "health_score": [100]*6,
    "temperature": np.random.normal(22,1,6),
    "power_usage": np.random.normal(120,10,6),
    "battery_level": np.random.normal(95,2,6)
})

iss_modules
```

```python
iss_modules["temperature"] += np.random.normal(0,0.2,6)
iss_modules["power_usage"] += np.random.normal(0,1,6)

iss_modules
```
