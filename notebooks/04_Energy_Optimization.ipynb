```python
import pandas as pd
import numpy as np

from scipy.optimize import minimize

import plotly.express as px
import plotly.graph_objects as go
```

```python
modules = pd.DataFrame({
    "module":[
        "Node1",
        "Node2",
        "Harmony",
        "Columbus",
        "Kibo",
        "Cupola"
    ],

    "power_usage":[
        120,
        110,
        140,
        150,
        130,
        90
    ]
})

modules
```

```python
current_consumption = modules["power_usage"].sum()

print(
    f"Current Consumption: {current_consumption} kW"
)
```

```python
minimum_power = np.array([
    100,
    90,
    120,
    130,
    110,
    80
])
```

```python
def objective(power):

    return np.sum(power)
```

```python
bounds = []

for min_power,current_power in zip(
    minimum_power,
    modules["power_usage"]
):

    bounds.append(
        (
            min_power,
            current_power
        )
    )
```

```python
result = minimize(
    objective,
    x0=modules["power_usage"],
    bounds=bounds
)
```

```python
optimized_power = result.x

modules["optimized_power"] = optimized_power

modules
```

```python
optimized_consumption = optimized_power.sum()

savings = (
    current_consumption
    -
    optimized_consumption
)

print(
    f"Current: {current_consumption:.1f} kW"
)

print(
    f"Optimized: {optimized_consumption:.1f} kW"
)

print(
    f"Savings: {savings:.1f} kW"
)
```

```python
fig = go.Figure()

fig.add_bar(
    name="Current",
    x=modules["module"],
    y=modules["power_usage"]
)

fig.add_bar(
    name="Optimized",
    x=modules["module"],
    y=modules["optimized_power"]
)

fig.update_layout(
    title="ISS Energy Optimization",
    barmode="group"
)

fig.show()
```

```python
kpi = pd.DataFrame({
    "Metric":[
        "Current Consumption",
        "Optimized Consumption",
        "Energy Savings"
    ],

    "Value":[
        current_consumption,
        optimized_consumption,
        savings
    ]
})

kpi
```
