```python
import pandas as pd
import numpy as np

from sklearn.ensemble import IsolationForest

import plotly.express as px
```

```python
n = 10000

df = pd.DataFrame({
    "temperature": np.random.normal(22,2,n),
    "power_usage": np.random.normal(120,15,n),
    "battery_level": np.random.normal(90,5,n),
    "vibration": np.random.normal(0.02,0.005,n)
})
```

```python
anomaly_indices = np.random.choice(
    df.index,
    size=100,
    replace=False
)

df.loc[anomaly_indices,"temperature"] = np.random.uniform(35,50,100)

df.loc[anomaly_indices,"vibration"] = np.random.uniform(0.05,0.10,100)
```

```python
features = [
    "temperature",
    "power_usage",
    "battery_level",
    "vibration"
]

model = IsolationForest(
    contamination=0.01,
    random_state=42
)

model.fit(df[features])
```

```python
df["anomaly"] = model.predict(df[features])
```

```python
df["anomaly"] = df["anomaly"].map({
    1:0,
    -1:1
})
```

```python
df["anomaly"].value_counts()
```

```python
fig = px.scatter(
    df,
    x="temperature",
    y="vibration",
    color="anomaly",
    title="ISS Anomaly Detection"
)

fig.show()
```

```python
anomalies = df[df["anomaly"] == 1]

anomalies.head(20)
```

```python
total_records = len(df)

anomaly_count = len(anomalies)

anomaly_rate = anomaly_count / total_records * 100

print(f"Total Records: {total_records}")
print(f"Anomalies: {anomaly_count}")
print(f"Anomaly Rate: {anomaly_rate:.2f}%")
```
