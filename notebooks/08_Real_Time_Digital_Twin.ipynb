```python
import pandas as pd
import numpy as np
import time

from IPython.display import clear_output
```

```python
iss_state = {

    "temperature":22,

    "power_usage":740,

    "battery_level":95,

    "vibration":0.02,

    "health_score":98

}
```

```python
def update_telemetry():

    iss_state["temperature"] += np.random.normal(0,0.2)

    iss_state["power_usage"] += np.random.normal(0,2)

    iss_state["battery_level"] += np.random.normal(0,0.1)

    iss_state["vibration"] += np.random.normal(0,0.001)

    iss_state["health_score"] += np.random.normal(0,0.05)

    return iss_state
```

```python
update_telemetry()
```

```python
for i in range(20):

    clear_output(wait=True)

    current_state = update_telemetry()

    print("ISS DIGITAL TWIN")

    print("---------------------")

    for key,value in current_state.items():

        print(f"{key}: {value:.2f}")

    time.sleep(1)
```

```python
history = []
```

```python
for i in range(100):

    current_state = update_telemetry()

    history.append(
        current_state.copy()
    )
```

```python
history_df = pd.DataFrame(history)

history_df.head()
```

```python
history_df.to_csv(
    "iss_telemetry.csv",
    index=False
)
```

```python
def check_alerts():

    alerts = []

    if iss_state["temperature"] > 26:

        alerts.append(
            "HIGH TEMPERATURE"
        )

    if iss_state["battery_level"] < 85:

        alerts.append(
            "LOW BATTERY"
        )

    if iss_state["vibration"] > 0.03:

        alerts.append(
            "HIGH VIBRATION"
        )

    return alerts
```

```python
alerts = check_alerts()

alerts
```

```python
def telemetry_report():

    return f"""
    ISS DIGITAL TWIN

    Temperature:
    {iss_state['temperature']:.2f}

    Power Usage:
    {iss_state['power_usage']:.2f}

    Battery:
    {iss_state['battery_level']:.2f}

    Health Score:
    {iss_state['health_score']:.2f}
    """
```
