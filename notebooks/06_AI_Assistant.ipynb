```python
digital_twin = {

    "health_score": 98,

    "current_power": 740,

    "optimized_power": 630,

    "energy_savings": 110,

    "predicted_failures": 3,

    "detected_anomalies": 12

}
```

```python
def ask_iss(question):

    question = question.lower()

    if "health" in question:
        return f"ISS Health Score: {digital_twin['health_score']}%"

    elif "power" in question:
        return f"Current Power Consumption: {digital_twin['current_power']} kW"

    elif "saving" in question:
        return f"Energy Savings: {digital_twin['energy_savings']} kW"

    elif "failure" in question:
        return f"Predicted Failures: {digital_twin['predicted_failures']}"

    elif "anomal" in question:
        return f"Detected Anomalies: {digital_twin['detected_anomalies']}"

    else:
        return "I don't understand the question."
```

```python
ask_iss(
    "What is the health score?"
)
```

```python
while True:

    question = input("ISS Assistant > ")

    if question.lower() == "exit":
        break

    print(
        ask_iss(question)
    )
```

```python
def generate_report():

    report = f"""
    ISS DIGITAL TWIN REPORT

    Health Score:
    {digital_twin['health_score']}%

    Current Power:
    {digital_twin['current_power']} kW

    Optimized Power:
    {digital_twin['optimized_power']} kW

    Energy Savings:
    {digital_twin['energy_savings']} kW

    Predicted Failures:
    {digital_twin['predicted_failures']}

    Detected Anomalies:
    {digital_twin['detected_anomalies']}
    """

    return report
```

```python
print(
    generate_report()
)
```
