```python
import pandas as pd
import numpy as np

import plotly.express as px
import plotly.graph_objects as go

from plotly.subplots import make_subplots
```

```python
health_score = 98

current_power = 740

optimized_power = 630

energy_savings = current_power - optimized_power

predicted_failures = 3

detected_anomalies = 12
```

```python
modules = pd.DataFrame({

    "module":[
        "Node1",
        "Node2",
        "Harmony",
        "Columbus",
        "Kibo",
        "Cupola"
    ],

    "health":[
        99,
        97,
        96,
        93,
        95,
        100
    ],

    "power":[
        120,
        110,
        140,
        150,
        130,
        90
    ]
})

modules
```

```python
fig = go.Figure(
    go.Indicator(
        mode="gauge+number",
        value=health_score,
        title={"text":"ISS Health Score"},
        gauge={
            "axis":{"range":[0,100]}
        }
    )
)

fig.show()
```

```python
fig = go.Figure()

fig.add_bar(
    x=["Current","Optimized"],
    y=[current_power, optimized_power]
)

fig.update_layout(
    title="Energy Consumption"
)

fig.show()
```

```python
fig = px.bar(
    modules,
    x="module",
    y="health",
    title="Module Health Status"
)

fig.show()
```

```python
failure_df = pd.DataFrame({

    "module":[
        "Battery 3",
        "Pump 1",
        "Solar Array"
    ],

    "probability":[
        82,
        61,
        34
    ]
})

fig = px.bar(
    failure_df,
    x="module",
    y="probability",
    title="Predicted Failures"
)

fig.show()
```

```python
kpi = pd.DataFrame({

    "Metric":[
        "Health Score",
        "Current Power",
        "Optimized Power",
        "Energy Savings",
        "Predicted Failures",
        "Detected Anomalies"
    ],

    "Value":[
        health_score,
        current_power,
        optimized_power,
        energy_savings,
        predicted_failures,
        detected_anomalies
    ]
})

kpi
```

```python
fig = make_subplots(
    rows=2,
    cols=2,

    specs=[
        [{"type":"indicator"},{"type":"bar"}],
        [{"type":"bar"},{"type":"bar"}]
    ],

    subplot_titles=(
        "Health Score",
        "Energy",
        "Module Health",
        "Failures"
    )
)

fig.add_trace(
    go.Indicator(
        mode="gauge+number",
        value=health_score
    ),
    row=1,
    col=1
)

fig.add_trace(
    go.Bar(
        x=["Current","Optimized"],
        y=[current_power, optimized_power]
    ),
    row=1,
    col=2
)

fig.add_trace(
    go.Bar(
        x=modules["module"],
        y=modules["health"]
    ),
    row=2,
    col=1
)

fig.add_trace(
    go.Bar(
        x=failure_df["module"],
        y=failure_df["probability"]
    ),
    row=2,
    col=2
)

fig.update_layout(
    height=900,
    title="ISS Digital Twin Executive Dashboard"
)

fig.show()
```

```python
dashboard_data = {

    "health_score": health_score,

    "current_power": current_power,

    "optimized_power": optimized_power,

    "energy_savings": energy_savings,

    "predicted_failures": predicted_failures,

    "detected_anomalies": detected_anomalies
}

dashboard_data
```
