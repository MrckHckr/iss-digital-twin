```python
import pandas as pd
import numpy as np

import plotly.graph_objects as go
```

```python
iss_modules = pd.DataFrame({

    "module":[
        "Node1",
        "Node2",
        "Harmony",
        "Columbus",
        "Kibo",
        "Cupola"
    ],

    "x":[
        0,
        2,
        1,
        -1,
        1,
        1
    ],

    "y":[
        0,
        0,
        1,
        1,
        -1,
        -2
    ],

    "z":[
        0,
        0,
        0,
        0,
        0,
        0
    ],

    "health":[
        99,
        97,
        95,
        92,
        96,
        100
    ]
})

iss_modules
```

```python
fig = go.Figure()

fig.add_trace(

    go.Scatter3d(

        x=iss_modules["x"],
        y=iss_modules["y"],
        z=iss_modules["z"],

        mode="markers+text",

        text=iss_modules["module"],

        marker=dict(

            size=15,

            color=iss_modules["health"],

            colorscale="Viridis",

            colorbar=dict(
                title="Health"
            )

        )
    )
)

fig.update_layout(

    title="ISS Digital Twin 3D",

    scene=dict(

        xaxis_title="X",

        yaxis_title="Y",

        zaxis_title="Z"

    )

)

fig.show()
```

```python
connections = [

    ("Node1","Harmony"),

    ("Harmony","Node2"),

    ("Harmony","Columbus"),

    ("Harmony","Kibo"),

    ("Kibo","Cupola")

]
```

```python
for start,end in connections:

    p1 = iss_modules[
        iss_modules["module"] == start
    ].iloc[0]

    p2 = iss_modules[
        iss_modules["module"] == end
    ].iloc[0]

    fig.add_trace(

        go.Scatter3d(

            x=[p1["x"],p2["x"]],
            y=[p1["y"],p2["y"]],
            z=[p1["z"],p2["z"]],

            mode="lines",

            showlegend=False

        )
    )

fig.show()
```

```python
iss_modules["hover"] = (

    "Module: "
    + iss_modules["module"]

    + "<br>Health: "
    + iss_modules["health"].astype(str)
    + "%"
)
```

```python
go.Scatter3d(

    x=iss_modules["x"],
    y=iss_modules["y"],
    z=iss_modules["z"],

    mode="markers+text",

    text=iss_modules["module"],

    hovertext=iss_modules["hover"],

    hoverinfo="text",

    marker=dict(
        size=15,
        color=iss_modules["health"],
        colorscale="Viridis"
    )

)
```

```python
def status_color(health):

    if health > 95:
        return "green"

    elif health > 90:
        return "yellow"

    else:
        return "red"
```

```python
iss_modules["status"] = (
    iss_modules["health"]
    .apply(status_color)
)

iss_modules
```

```python
iss_modules.to_csv(
    "iss_modules.csv",
    index=False
)
```
